```python
# declare name of the collection
DATA_SET    = "beomi/KoAlpaca-v1.1a"
DATA_FILE   = "./data/KoAlpaca-train.csv"
TEXT_COLUMN = "instruction"
INDEX_NAME = "MyDocuments"
```

```python
import pandas as pd
from datasets import load_dataset

load_data = load_dataset(DATA_SET, split="train")
load_data.to_csv(DATA_FILE)
df = pd.read_csv(DATA_FILE)
metadata_df = df.drop(columns=[TEXT_COLUMN])  # 메타데이터 컬럼들

df.head(5)
metadata_df.head(5)
```

```python
import weaviate
from weaviate_toolkit.WeaviateManager import WeaviateManager
from weaviate_toolkit.EmbeddingModelSetup import setup_bge_m3_embeddings as setup_embeddings


manager = WeaviateManager(
    weaviate_client=weaviate.connect_to_local(),
    embeddings=setup_embeddings()
)
```

```python
manager.delete_index(INDEX_NAME)
```

```python
manager.load_dataframe_to_weaviate(
    df=df[:100],
    text_column=TEXT_COLUMN,  # 벡터화할 텍스트 컬럼
    index_name=INDEX_NAME,
    metadata_columns=list(df.columns) # 메타데이터 컬럼들
)
```

```python
manager.search_similar("거짓말 탐지기", k=2)
```
